## 4. Xác thực Dữ liệu FewRel & Sampler Integrity (Preflight Check)


```python
import sys
import json
import torch
import numpy as np
import inspect
from types import ModuleType

# 1. Vá lỗi tương thích ragged array cho NumPy hiện đại (>= 1.24)
# Keep the true NumPy function across cell re-runs; otherwise wrappers recurse.
if not hasattr(np, "_conpl_original_asarray"):
    np._conpl_original_asarray = np.asarray
_orig_asarray = np._conpl_original_asarray
def _safe_asarray(a, *args, **kwargs):
    try:
        return _orig_asarray(a, *args, **kwargs)
    except ValueError:
        return _orig_asarray(a, dtype=object, *args, **kwargs)
np.asarray = _safe_asarray

# 2. Vá lỗi tương thích cho huggingface_hub hiện đại (>= 1.0)
try:
    import huggingface_hub
    import huggingface_hub.file_download
    from tqdm.auto import tqdm

    # Vá HfFolder
    if not hasattr(huggingface_hub, 'HfFolder'):
        class HfFolder:
            @staticmethod
            def get_token():
                return getattr(huggingface_hub, 'get_token', lambda: None)()
            @staticmethod
            def save_token(token):
                pass
        huggingface_hub.HfFolder = HfFolder

    # Vá tqdm trong file_download để không lỗi AttributeError
    if not hasattr(huggingface_hub.file_download, 'tqdm'):
        huggingface_hub.file_download.tqdm = tqdm

    # Vá tham số use_auth_token thành token cho hf_hub_download
    if 'token' in inspect.signature(huggingface_hub.hf_hub_download).parameters:
        _orig_download = huggingface_hub.hf_hub_download
        def _compat_download(*args, **kwargs):
            if 'use_auth_token' in kwargs:
                tok = kwargs.pop('use_auth_token')
                if tok not in (None, False, True):
                    kwargs['token'] = tok
            return _orig_download(*args, **kwargs)
        huggingface_hub.hf_hub_download = _compat_download
except Exception:
    pass

# 3. Bỏ qua runtime dependency version assertion cũ của transformers trên Python 3.13
dummy_check = ModuleType('transformers.dependency_versions_check')
dummy_check.dep_version_check = lambda *a, **k: None
sys.modules['transformers.dependency_versions_check'] = dummy_check

# 4. Đảm bảo import ưu tiên từ thư mục ConPL hiện tại
if '.' not in sys.path:
    sys.path.insert(0, '.')

from encode import BERTSentenceEncoderPrompt
from dataprocess import data_sampler_bert_prompt_deal_first_task

# Patch the exact function reference used by legacy Transformers cached_file().
import transformers.utils.hub as _transformers_hub
_hf_download_impl = _transformers_hub.hf_hub_download
def _hub_download_compat(*args, **kwargs):
    if 'use_auth_token' in kwargs:
        kwargs['token'] = kwargs.pop('use_auth_token')
    return _hf_download_impl(*args, **kwargs)
_transformers_hub.hf_hub_download = _hub_download_compat

with open('config/config_fewrel_5and10.json', 'r') as f:
    cfg = json.load(f)

cfg['rel_cluster_label'] = 'data/fewrel/CFRLdata_10_100_10_5/rel_cluster_label_0.npy'
cfg['training_file'] = 'data/fewrel/CFRLdata_10_100_10_5/train_0.txt'
cfg['valid_file'] = 'data/fewrel/CFRLdata_10_100_10_5/valid_0.txt'
cfg['test_file'] = 'data/fewrel/CFRLdata_10_100_10_5/test_0.txt'
cfg['first_task_k-way'] = 10
cfg['k-shot'] = 5
cfg['device'] = 'cuda' if torch.cuda.is_available() else 'cpu'

print('Đang nạp BERT Prompt Sentence Encoder và khởi tạo Sampler...')
encoder = BERTSentenceEncoderPrompt(cfg)
sampler = data_sampler_bert_prompt_deal_first_task(cfg, encoder.tokenizer, 'e1 mask e2 . ')
sampler.set_seed(100)

print('=' * 60)
print('XÁC THỰC DỮ LIỆU & ENCODER THÀNH CÔNG')
print('=' * 60)
print(f'✓ Thiết bị sử dụng: {cfg["device"]}')
print(f'✓ Số lượng quan hệ (Relations): {len(sampler.id2rel) - 1}')
print(f'✓ Kích thước từ vựng (Vocab size): {len(encoder.tokenizer)}')
print('✓ Data Sampler khởi tạo hoàn toàn chính xác!')
```

## 6. Phân tích Kết quả & Đối chiếu Paper Table 1 (ACL 2023)


```python
import os
import json
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

paper_reported = {
    'T1': 95.72, 'T2': 93.53, 'T3': 91.31, 'T4': 89.95,
    'T5': 88.93, 'T6': 88.39, 'T7': 87.43, 'T8': 85.77
}

output_dir = Path(os.environ.get('CONPL_OUTPUT_DIR', '../../results/conpl/local_t4_run_6seq'))
metrics_files = sorted(output_dir.glob('seed_*/task_metrics.jsonl'))

print('=' * 60)
print('TỔNG HỢP KẾT QUẢ THỰC NGHIỆM CONPL')
print('=' * 60)

if metrics_files:
    seq_accs = {}
    for mf in metrics_files:
        seed_name = mf.parent.name
        with open(mf, 'r', encoding='utf-8') as f:
            records = [json.loads(line) for line in f]
        seq_accs[seed_name] = {f"T{r['task']}": r['whole_acc'] * 100 for r in records}
    
    df = pd.DataFrame(seq_accs).T
    df.loc['Local T4 Mean'] = df.mean()
    df.loc['Paper Table 1'] = pd.Series(paper_reported)
    if 'Local T4 Mean' in df.index and 'Paper Table 1' in df.index:
        df.loc['Diff (Local - Paper)'] = df.loc['Local T4 Mean'] - df.loc['Paper Table 1']
    
    display(df.round(2))

    # Vẽ biểu đồ đường cong độ chính xác tích lũy
    plt.figure(figsize=(9, 5))
    tasks = list(paper_reported.keys())
    plt.plot(tasks, [paper_reported[t] for t in tasks], 'o--', color='#d9534f', label='Paper Reported (Table 1 Mean)')
    
    local_means = [df.loc['Local T4 Mean'].get(t, None) for t in tasks]
    valid_tasks = [t for t, v in zip(tasks, local_means) if pd.notnull(v)]
    valid_vals = [v for v in local_means if pd.notnull(v)]
    if valid_vals:
        plt.plot(valid_tasks, valid_vals, 's-', color='#2e6da4', linewidth=2, label='Local T4 Mean')
    
    plt.xlabel('Continual Tasks', fontsize=12)
    plt.ylabel('Cumulative Whole Accuracy (%)', fontsize=12)
    plt.title('ConPL FewRel 5-shot: Local T4 vs. Paper Table 1', fontsize=13, fontweight='bold')
    plt.grid(True, linestyle=':', alpha=0.6)
    plt.ylim(80, 100)
    plt.legend(fontsize=11)
    plt.tight_layout()
    plot_file = output_dir / 'accuracy_comparison.png'
    plt.savefig(plot_file, dpi=200)
    plt.show()
    print(f'✓ Đã lưu biểu đồ đối chiếu tại: {plot_file}')
else:
    print('Chưa có file task_metrics.jsonl được tạo. Giá trị bài báo công bố:')
    for k, v in paper_reported.items():
        print(f'{k}: Paper={v:.2f}%')
```
